## Pandas Scrape
Another project on how to scrape a Wikipedia table using pandas and requests libraries
These data can be used for data analysis, machine learning model training, and many more

```python
# Import the libraries to perform the task
import requests as r # This libraries act as API which connect the website to the project
import pandas as pd # This one convert raw HTML data into a DataFrame and export them to an Excel file
```

```python
# This URL is the URL of the website we want to scrape
url = "https://en.wikipedia.org/wiki/List_of_largest_banks"
# Headers prevent the website from blocking our project
headers = {"User-Agent":"Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/152.0.0.0 Safari/537.36"}
```

```python
# Now this is the fetching of the data using requests. get method to retrieve the website content
response = r.get(url, headers = headers)
```

```python
# Here we convert the raw HTML table into a pandas DataFrame
df = pd.read_html(response.text)
```

```python
# Here we extract the first table and assign it to a variable called table
table = df[0]
```

```python
# Now we can export our table to an excel file
table.to_excel("pandas_data.xlsx", index=False)
```

```python
# I scrape another table in the same URL
table1 = df[1]
```
